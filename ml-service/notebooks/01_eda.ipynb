# EDA — Gym Members Exercise Tracking

Notebook ini mendokumentasikan eksplorasi awal dataset yang dipakai GymSense.
Jalankan dari folder `ml-service` agar path relatif bekerja.

- Target: `Experience_Level` (1 = Beginner, 2 = Intermediate, 3 = Advanced)
- 14 fitur: 12 numerik + 2 kategorikal
- Split training dan evaluasi ada di `train.py`

```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
plt.style.use('ggplot')

df = pd.read_csv('../data/raw/gym_members_exercise_tracking.csv')
df.shape
```

```python
df.head()
```

```python
df.info()
print('Missing values:', int(df.isna().sum().sum()))
print('Duplicate rows:', int(df.duplicated().sum()))
```

```python
df['Experience_Level'].map({1: 'Beginner', 2: 'Intermediate', 3: 'Advanced'}).value_counts()
```

```python
df.select_dtypes('number').describe().T
```

```python
numeric = df.select_dtypes('number').drop(columns=['Experience_Level'])
corr = df.select_dtypes('number').corr()['Experience_Level'].drop('Experience_Level').sort_values()
corr.plot(kind='barh', figsize=(8, 6), title='Korelasi fitur numerik dengan Experience_Level')
plt.tight_layout()
plt.show()
```

```python
# Outlier kasar dengan aturan IQR
for column in numeric.columns:
    q1, q3 = numeric[column].quantile([0.25, 0.75])
    iqr = q3 - q1
    count = int(((numeric[column] < q1 - 1.5 * iqr) | (numeric[column] > q3 + 1.5 * iqr)).sum())
    print(f'{column:<32} {count} outlier')
```

```python
df.pivot(columns='Experience_Level', values='Workout_Frequency (days/week)').plot(
    kind='box', figsize=(8, 5), grid=True
)
plt.title('Frekuensi latihan per level pengalaman')
plt.xlabel('Experience_Level')
plt.ylabel('Workout_Frequency (days/week)')
plt.show()
```
